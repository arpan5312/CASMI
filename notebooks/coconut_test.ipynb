```python
import pandas as pd
import sys
import os


sys.path.append(os.path.abspath(".."))

from src.candidate_generation.ms1 import append_neutral_mass,data_frame_modifier,find_candidates
test_path = "../data/test.parquet"
db_path = "../data/coconutdb.csv"
df = pd.read_parquet(test_path,columns=['adduct','precursor_mz'])
coconut = pd.read_csv(db_path)

ndf = append_neutral_mass(df)

mdf = data_frame_modifier(ndf)

candidates = find_candidates(mdf,246.100376)


for _, candidate in candidates.iterrows():
    
    mass = candidate["neutral_mass"]
    
    tolerance = float(mass) * 6e-6
    
    lower = mass - tolerance
    upper = mass + tolerance
    
    hits = coconut[
        (coconut["exact_molecular_weight"] >= lower) &
        (coconut["exact_molecular_weight"] <= upper)
    ]

print(hits)
```
